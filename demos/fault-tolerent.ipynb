```python
import sys; sys.path.insert(0,'..')
import pyzx as zx
from pyzx import VertexType, EdgeType
```

```python
g = zx.Graph()

i = g.add_vertex(VertexType.BOUNDARY, 0, 0)
v = g.add_vertex(VertexType.Z, 0, 1, 0)
w = g.add_vertex(VertexType.X, 0, 2, 0)
o = g.add_vertex(VertexType.Z, 0, 3)

g.add_edge((i,v), EdgeType.SIMPLE)
g.add_edge((v,w), EdgeType.SIMPLE)

g.set_noise_model("edge_flip") # enable noise model

g.noise_model.set_idealized((v,w)) # set edge (v,w) to ideal.

g.add_edge((w,o), EdgeType.SIMPLE) # edges are non-ideal by default
```

```python
from pyzx.utils import settings
settings.drawing_backend = "d3"

display(zx.draw(g))
```

```python
settings.drawing_backend = "matplotlib"

display(zx.draw(g))
```

```python
g.to_tikz()
```

```python
g.set_noise_model(None) # removes noise model

display(zx.draw(g))
```
